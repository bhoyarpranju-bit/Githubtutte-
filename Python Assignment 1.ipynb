```python
def calculate_area(radius):
    pi = 3.14
    area = pi * radius ** 2
    return area

# Test the function with different radius values
print("Area for radius 5:", calculate_area(5))
print("Area for radius 10:", calculate_area(10))
print("Area for radius 7.5:", calculate_area(7.5))
```

```python
def check_palindrome(text: str) -> bool:
    """
    Checks if a given string is a palindrome.
    Ignores casing, spaces, and punctuation.
    """
    # Clean the string: convert to lowercase and keep only alphanumeric characters
    cleaned_text = "".join(char.lower() for char in text if char.isalnum())
    
    # Compare the cleaned string with its reverse
    return cleaned_text == cleaned_text[::-1]

# --- Testing the function with different strings ---

test_strings = [
    "radar",                # Standard lowercase palindrome
    "Hello",                # Not a palindrome
    "A man, a plan, a canal: Panama", # Palindrome with mixed case, spaces, and punctuation
    "12321",                # Numeric palindrome
    "racecar",              # Standard lowercase palindrome
    "Python"                # Not a palindrome
]

print("--- Palindrome Test Results ---")
for text in test_strings:
    result = check_palindrome(text)
    print(f"'{text}' -> {result}")
    
```

```python
def count_vowels(input_string):
    """
    Counts the number of vowels (a, e, i, o, u) in the input string.
    """
    vowels = "aeiou"
    count = 0
    
    # Loop through each character in the string
    for char in input_string.lower():
        if char in vowels:
            count += 1
            
    return count
# --- Testing the function ---

# Test Case 1: Standard lowercase string
test1 = "hello world"
print(f"String: '{test1}' -> Vowel Count: {count_vowels(test1)}")

# Test Case 2: Mixed case string
test2 = "Python Programming IS Fun"
print(f"String: '{test2}' -> Vowel Count: {count_vowels(test2)}")

# Test Case 3: No vowels present
test3 = "rhythm fly"
print(f"String: '{test3}' -> Vowel Count: {count_vowels(test3)}")

# Test Case 4: Empty string
test4 = ""
print(f"String: '{test4}' -> Vowel Count: {count_vowels(test4)}")
```

```python
def calculate_factorial(n):
    """
    Calculates the factorial of a non-negative integer n iteratively.
    """
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    
    result = 1
    for i in range(1, n + 1):
        result *= i
        
    return result

# --- Testing the function ---

test_numbers = [0, 1, 5, 10]

for num in test_numbers:
    print(f"The factorial of {num} is: {calculate_factorial(num)}")
```

```python
def check_prime(n):
    """
    Checks if an integer n is a prime number.
    Returns True if prime, False otherwise.
    """
    # Prime numbers must be greater than 1
    if n <= 1:
        return False
    
    # 2 is the only even prime number
    if n == 2:
        return True
        
    # Exclude all other even numbers
    if n % 2 == 0:
        return False
        
    # Check odd factors up to the square root of n
    # (Any composite number must have a factor less than or equal to its square root)
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
            
    return True

# --- Testing the function ---

test_numbers = [1, 2, 4, 17, 20, 29, 97, 100]

for num in test_numbers:
    result = "a Prime number" if check_prime(num) else "NOT a Prime number"
    print(f"The number {num} is {result}.")
```

```python
# Lambda function: reverses the string using slicing [::-1] after lowering and stripping spaces
is_palindrome = lambda s: (clean := s.lower().replace(" ", "")) == clean[::-1]

# --- Testing the function ---

test_strings = [
    "radar",          # Standard palindrome
    "Race Car",       # Palindrome with mixed casing and spaces
    "hello",          # Non-palindrome
    "A man a plan a canal Panama", # Classic phrase palindrome
    "12321",          # Numeric palindrome
    ""                # Empty string (technically reads the same forward and backward)
]

for text in test_strings:
    result = "a Palindrome" if is_palindrome(text) else "NOT a Palindrome"
    print(f"'{text}' is {result}.")
```

```python
# Lambda function: multiplies length (l) by width (w)
calculate_rectangle_area = lambda l, w: l * w

# --- Testing the function ---

test_cases = [
    (5, 10),      # Standard integers
    (3.5, 7.2),   # Floating-point numbers
    (12, 12),     # Square (special case of a rectangle)
    (0, 15)       # Zero boundary case
]

for length, width in test_cases:
    area = calculate_rectangle_area(length, width)
    print(f"Length: {length}, Width: {width} -> Area: {area}")
```

```python
# Lambda function: multiplies Celsius (c) by 1.8 (or 9/5) and adds 32
celsius_to_fahrenheit = lambda c: (c * 9/5) + 32

# --- Testing the function ---

test_temperatures = [
    0,      # Freezing point of water
    100,    # Boiling point of water
    -40,    # The temperature where Celsius and Fahrenheit are equal
    22.5,   # Typical room temperature
    37      # Normal human body temperature
]

for celsius in test_temperatures:
    fahrenheit = celsius_to_fahrenheit(celsius)
    print(f"{celsius}°C is equal to {fahrenheit}°F")
```

```python
def reverse_string_slicing(input_string):
    """
    Reverses the given string using Python slicing syntax [:: -1].
    """
    # [::-1] starts from the end of the string and moves backward with a step of -1
    return input_string[::-1]

# --- Testing the program ---

test_strings = [
    "hello",         # Standard word
    "Python 3.12",   # String with numbers and spaces
    "racecar",       # Palindrome (should look identical reversed)
    "A",             # Single character
    ""               # Empty string
]

for text in test_strings:
    reversed_text = reverse_string_slicing(text)
    print(f"Original: '{text}' -> Reversed: '{reversed_text}'")
```

```python
import string

def count_word_occurrences(text, target_word):
    """
    Counts the occurrences of a specific word in a given string,
    ignoring punctuation and case sensitivity.
    """
    # Convert text and target word to lowercase for case-insensitive matching
    text_lower = text.lower()
    target_lower = target_word.lower()
    
    # Remove punctuation marks from the text
    # string.punctuation contains characters like !, ., ?, ,, etc.
    cleaned_text = text_lower.translate(str.maketrans("", "", string.punctuation))
    
    # Split the string into a list of individual words
    words_list = cleaned_text.split()
    
    # Count and return the occurrences of the target word
    return words_list.count(target_lower)

# --- Testing the program ---

test_cases = [
    {
        "text": "Python is a great programming language. Python is versatile and powerful.",
        "word": "Python"
    },
    {
        "text": "The cat sat on the mat, but another cat was outside.",
        "word": "cat"
    },
    {
        "text": "Hello world! Welcome to the world of programming.",
        "word": "world"  # Tests if punctuation removal works (matching 'world!' and 'world')
    },
    {
        "text": "Red leather, yellow leather, red leather, yellow leather.",
        "word": "blue"   # Tests a word that does not exist in the text
    }
]

for case in test_cases:
    count = count_word_occurrences(case["text"], case["word"])
    print(f"Text: \"{case['text']}\"")
    print(f"Word to count: '{case['word']}' -> Occurrences: {count}\n")
```

```python
import re

def remove_all_whitespace(input_string):
    """
    Removes all whitespace characters from the given string using regular expressions.
    """
    # \s matches any whitespace character (space, tab \t, newline \n, etc.)
    # The empty string "" replaces all matches
    return re.sub(r'\s+', '', input_string)

# --- Testing the program ---

test_strings = [
    "Hello World",                         # Standard single space
    "  Python   Programming  ",            # Leading, trailing, and multiple spaces
    "Line1\nLine2\tTabbed\rCarriage",       # Newlines, tabs, and carriage returns
    "No_whitespace_here",                  # No whitespaces present
    "   "                                  # Only whitespaces
]

for text in test_strings:
    cleaned_text = remove_all_whitespace(text)
    # repr() is used to display the raw characters (like \n or \t) in the original string
    print(f"Original: {repr(text)}")
    print(f"Cleaned : {repr(cleaned_text)}\n")
    
```
