```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
```

```python
df.head()
```

```python
# Display the first 5 rows
print(df.head())
```

```python
# Display rows and columns count
print(df.shape)
```

```python
# Get rows and columns separately
rows, columns = df.shape
print(f"Rows: {rows}, Columns: {columns}")
```

```python
# Display data types of each column
print(df.dtypes)
```

```python
# Drop the 'Cabin' column and save changes back to 'df'
df = df.drop(columns=['Cabin'])
```

```python
df.head()
```

```python
# Calculate the mean age
mean_age = df['Age'].mean()
```

```python
# Fill missing values in the 'Age' column with the mean age
df['Age'] = df['Age'].fillna(mean_age)
```

```python
df.head()
```

```python
# Drop rows where 'Embarked' has missing values and reassign to df
df = df.dropna(subset=['Embarked'])
# Alternative using inplace=True:
# df.dropna(subset=['Embarked'], inplace=True)
```

```python
df.head()
```

```python
# Calculate overall survival rate (as a decimal)
survival_rate = df['Survived'].mean()
```

```python
# Convert to a percentage for easier reading
survival_percentage = survival_rate * 100
print(f"Overall Survival Rate: {survival_percentage:.2f}%")
```

```python
# Calculate survival rates grouped by sex
survival_by_sex = df.groupby('Sex')['Survived'].mean()

# Display the results as percentages
for sex, rate in survival_by_sex.items():
    print(f"Survival rate for {sex}: {rate * 100:.2f}%")
```

```python
# Calculate average age grouped by sex
average_age_by_sex = df.groupby('Sex')['Age'].mean()

# Display the results
for sex, avg_age in average_age_by_sex.items():
    print(f"Average age for {sex}: {avg_age:.1f} years")
```

```python
# Calculate survival rates grouped by passenger class
survival_by_class = df.groupby('Pclass')['Survived'].mean()

# Display the results as percentages
for pclass, rate in survival_by_class.items():
    print(f"Survival rate for Class {pclass}: {rate * 100:.2f}%")
```

```python
# Calculate survival rates grouped by port of embarkation
survival_by_embarkation = df.groupby('Embarked')['Survived'].mean()

# Display the results as percentages
for port, rate in survival_by_embarkation.items():
    # C = Cherbourg, Q = Queenstown, S = Southampton
    print(f"Survival rate for Port {port}: {rate * 100:.2f}%")
```

```python
# Create a cross-tabulation table for raw counts of survivors
# Setting df['Survived'] == 1 filters for survivors only
survivors_df = df[df['Survived'] == 1]
ct_table = pd.crosstab(index=survivors_df['Pclass'], columns=survivors_df['Sex'])
print(ct_table)
```

```python
# 1. Recreate the cross-tabulation from the previous step
survivors_df = df[df['Survived'] == 1]
ct_table = pd.crosstab(index=survivors_df['Pclass'], columns=survivors_df['Sex'])

# 2. Stack the table
stacked_table = ct_table.stack()

print(stacked_table)
```

```python
# Assuming 'stacked_table' is the multi-indexed Series from the previous step
restored_table = stacked_table.unstack()

print(restored_table)
```

```python
# 1. Recreate your cross-tabulation and stack it
survivors_df = df[df['Survived'] == 1]
ct_table = pd.crosstab(index=survivors_df['Pclass'], columns=survivors_df['Sex'])
stacked_table = ct_table.stack()

# 2. Convert the stacked Series into a flat DataFrame so it can be merged
# 'name' gives a column title to the counted values
stacked_df = stacked_table.reset_index(name='Class_Sex_Survivor_Count')

# 3. Merge the original DataFrame with the flattened stacked table
# We match on 'Pclass' and 'Sex' to link the counts to the right rows
merged_df = df.merge(stacked_df, on=['Pclass', 'Sex'], how='left')

print(merged_df.head())
```

```python
# Calculate average age per class
avg_age_by_class = df.groupby('Pclass')['Age'].mean()
print("--- Average Age by Pclass ---")
print(avg_age_by_class)
```

```python
# Calculate min, max, and average survival rates per class
survival_stats = df.groupby('Pclass')['Survived'].agg(['min', 'max', 'mean'])

# Rename columns for clarity
survival_stats.columns = ['Min_Survived', 'Max_Survived', 'Survival_Rate']
print("\n--- Survival Stats by Pclass ---")
print(survival_stats)
```

```python
df = pd.read_csv("data.csv")

# Create a mapping dictionary
# (Note: In this specific case, the keys and values match perfectly, 
# which acts as a quick way to explicitly cast or clean the categories)
embarkation_map = {
    'S': 'S',
    'C': 'C',
    'Q': 'Q'
}

# Apply the mapping function back to the column
df['Embarked'] = df['Embarked'].map(embarkation_map)

print(df['Embarked'].value_counts())
```

```python
# Use apply to measure character length of each name
df['Name_Length'] = df['Name'].apply(len)

# Display the first few rows to verify the new column
print(df[['Name', 'Name_Length']].head())
```
