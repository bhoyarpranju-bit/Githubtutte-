```python
import pandas as pd
import numpy as np
```

```python
# Read csv file
df = pd.read_csv('titanic.csv')
```

```python
# Display first 5 rows
df.head()
```

```python
print(f"Number of rows: {len(df)}")
print(f"Number of columns: {len(df.columns)}")
```

```python
# 1. Summary statistics for numerical columns
print("--- Numerical Summary Statistics ---")
print(df.describe())
print("\n")
```

```python
# 2. Summary statistics for categorical (text) columns
print("--- Categorical Summary Statistics ---")
print(df.describe(include='object'))
```

```python
# Identify columns with missing values 
print("--- Missing Values Per Column ---")
missing_counts = df.isnull().sum()
print(missing_counts)
```

```python
# Filter to show only columns that actually have missing values
columns_with_missing = missing_counts[missing_counts > 0]
print(f"\nColumns containing missing data: {list(columns_with_missing.index)}\n")
```

```python
# Impute missing values in 'Age' with the Mean Age ---
# Calculate the mean age (Pandas automatically ignores NaN values when calculating the mean)
mean_age = df['Age'].mean()
print(f"Calculated Mean Age: {mean_age:.2f}")
```

```python
# Fill the missing values in the 'Age' column
df['Age'] = df['Age'].fillna(mean_age)
# Verify the Results 
print("\n--- Dataset After Imputation ---")
print(df)

print("\nRemaining missing values in 'Age':", df['Age'].isnull().sum())
```

```python
df.head()
```

```python
print("\nMissing values in 'Embarked':", df['Embarked'].isnull().sum())
```

```python
#Find the most frequent value (mode) 
# .mode() returns a Series, so we grab the first element [0]
most_frequent_embarked = df['Embarked'].mode()[0]
print(f"\nMost frequent 'Embarked' value: '{most_frequent_embarked}'")
```

```python
# Impute the missing values 
df['Embarked'] = df['Embarked'].fillna(most_frequent_embarked)
# Verify the Results 
print("\n--- After Imputation ---")
print(df)
print("\nMissing values in 'Embarked' now:", df['Embarked'].isnull().sum())
```

```python
df.head()
```

```python
df['Sex'] = df['Sex'].map({'female': 0, 'male': 1})
print(df)
```

```python
numeric_cols = ['Age', 'Fare', 'SibSp', 'Parch']

df[numeric_cols].describe()
```

```python
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"{col}: {len(outliers)} outliers")
```

```python
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)
```

```python
# Check duplicates
print("Number of duplicate rows:", df.duplicated().sum())

# Display duplicates
print(df[df.duplicated()])

# Remove duplicates
df = df.drop_duplicates()

# Verify
print("Number of duplicate rows after removal:", df.duplicated().sum())
```
